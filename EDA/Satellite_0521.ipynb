```python
import requests

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

# 시도할 URL 목록
urls = [
    "https://celestrak.org/pub/TLE/active.csv",
    "https://celestrak.org/SOCRATES/query.php?CATNR=ALL&FORMAT=csv",
    "https://celestrak.org/pub/satcat.csv",
]

for url in urls:
    try:
        response = requests.get(url, headers=headers)
        print(f"[{response.status_code}] {url}")
    except Exception as e:
        print(f"[ERROR] {url} → {e}")
```

```python
import requests
import os

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

url = "https://celestrak.org/NORAD/elements/gp.php?GROUP=active&FORMAT=csv"

print("다운로드 중...")
response = requests.get(url, headers=headers)
response.raise_for_status()

# 저장 경로 (원하는 경로로 변경 가능)
save_path = "./data/active_satellites.csv"
os.makedirs(os.path.dirname(save_path), exist_ok=True)

with open(save_path, "w", encoding="utf-8") as f:
    f.write(response.text)

print(f"저장 완료: {save_path}")
print(f"총 위성 수: {len(response.text.splitlines()) - 1}개")  # 헤더 제외
```

```python
import requests
import os

# 브라우저 요청과 최대한 동일하게 설정
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Referer": "https://celestrak.org/NORAD/elements/table.php?GROUP=active&FORMAT=csv",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "Accept-Language": "ko-KR,ko;q=0.9,en-US;q=0.8,en;q=0.7",
    "Accept-Encoding": "gzip, deflate, br",
    "Connection": "keep-alive",
}

# Session 사용 (쿠키 유지)
session = requests.Session()

# 먼저 메인 페이지 방문 (쿠키 획득)
session.get("https://celestrak.org/NORAD/elements/table.php?GROUP=active&FORMAT=csv", headers=headers)

# 실제 데이터 요청
url = "https://celestrak.org/NORAD/elements/gp.php?GROUP=active&FORMAT=csv"
response = session.get(url, headers=headers)

print(f"상태코드: {response.status_code}")
print(response.text[:300])  # 앞부분 확인
```
