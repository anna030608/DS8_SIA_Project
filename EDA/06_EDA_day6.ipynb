```python
import folium
import pandas as pd

# 1. 데이터 구성
data = [
    ["Pingtan", 25.50, 119.79, "상륙훈련 지역", "중국 상륙훈련 핵심"],
    ["Xiamen", 24.48, 118.09, "항만·병참", "대만 최단거리 병참"],
    ["Longtian AB", 25.67, 119.38, "공군기지", "ADIZ 작전 핵심"],
    ["Hui'an AB", 25.03, 118.80, "공군기지", "대만 접근 공군"],
    ["Quanzhou Port", 24.88, 118.68, "항만·해군", "봉쇄 지원"],
    ["Ningde Naval Base", 26.67, 119.55, "해군기지", "동부전구 해군 핵심"],
    ["Taichung AB", 24.26, 120.62, "대만 공군기지", "중부 방공 대응"],
    ["Hualien AB", 24.02, 121.62, "대만 공군기지", "생존성 핵심"],
    ["Chihhang AB", 22.79, 121.18, "대만 공군기지", "후방 공군 유지"],
    ["Bashi Channel", 20.70, 121.90, "해협 chokepoint", "남부 우회 접근"],
    ["Miyako Strait", 24.80, 125.30, "해협 chokepoint", "태평양 진출 핵심"],
    ["Taiwan Strait Median Line", 24.00, 119.50, "전략 해역", "회색지대 분석"],
    ["Yulin Naval Base", 18.22, 109.69, "해군기지", "장거리 해군작전"]
]

df = pd.DataFrame(data, columns=["Name", "Lat", "Lon", "Type", "Value"])

# 2. 지도 초기화 (타이완 해협 중심으로 설정)
m = folium.Map(location=[23.5, 120.5], zoom_start=6, tiles="CartoDB positron")

# 3. 마커 추가
for i, row in df.iterrows():
    # 기지 유형에 따라 색상 변경 (중국: 빨강, 대만: 파랑, 해역: 검정)
    if "대만" in row['Type']:
        color = 'blue'
    elif "해협" in row['Type'] or "해역" in row['Type']:
        color = 'black'
    else:
        color = 'red'
        
    folium.Marker(
        location=[row['Lat'], row['Lon']],
        popup=f"<b>{row['Name']}</b><br>대상: {row['Type']}<br>가치: {row['Value']}",
        tooltip=row['Name'],
        icon=folium.Icon(color=color, icon='info-sign')
    ).add_to(m)

# 4. 지도 저장 및 출력
m.save("geoint_map.html")
m
```
