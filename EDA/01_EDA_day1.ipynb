```python
import pandas_gbq

query = """
SELECT SQLDATE, EventCode, GoldsteinScale, NumMentions, ActionGeo_Lat, ActionGeo_Long
FROM `gdelt-bq.full.events`
WHERE SQLDATE >= 20130401
  AND (
    (Actor1CountryCode = 'CHN' AND Actor2CountryCode = 'TWN') OR 
    (Actor1CountryCode = 'TWN' AND Actor2CountryCode = 'CHN')
  )
  AND IsRootEvent = 1
"""

# 프로젝트 ID 입력 후 바로 Pandas DataFrame으로 변환
project_id = "gdelt-analysis-494301"
df = pandas_gbq.read_gbq(query, project_id=project_id)

print(df.head())
```

```python
df['SQLDATE'] = pd.to_datetime(df['SQLDATE'].astype(str), format='%Y%m%d')
```

```python
# 데이터의 기본 정보 요약
print("--- [데이터 구조 확인] ---")
print(df.info())

# 결측치(NaN) 확인 - 리포트 신뢰도와 직결
print("\n--- [결측치 비율 (%)] ---")
print(df.isnull().mean() * 100)

# 주요 수치 데이터의 기초 통계량 (평균, 최소, 최대, 4분위수)
print("\n--- [주요 수치 통계] ---")
display(df[['GoldsteinScale']].describe())
```

```python
# 숫자를 문자열로 바꾼 뒤, 
# 3자리 미만인 경우(예: 1~99)는 그대로 쓰고, 3자리 이상인 경우 앞의 2자리를 추출
df['RootCode'] = df['EventCode'].astype(str).apply(lambda x: x.zfill(2) if len(x) <= 2 else x[:2])

# 또는 GDELT 표준 형식을 따를 경우 (3~4자리 코드에서 앞 2자리 추출)
df['RootCode'] = df['EventCode'].astype(str).str.zfill(3).apply(lambda x: x[:2] if len(x) == 3 else x[:2])

# 코드별 빈도수 계산 및 정렬
code_counts = df['RootCode'].value_counts().sort_index()

print("--- CAMEO 대분류별 데이터 수 ---")
print(code_counts)
```

```python
# 1. RootCode(대분류) 추출
df['RootCode'] = df['EventCode'].astype(str).str.zfill(3).str[:2]

# 2. 코드별 사건 수(Row Count)와 기사 언급량(NumMentions Sum) 계산
comparison_df = df.groupby('RootCode').agg({
    'EventCode': 'count',      # 사건의 수 (데이터 행의 개수)
    'NumMentions': 'sum'       # 기사의 총합 (데이터 속 언급 횟수의 합)
}).rename(columns={'EventCode': 'Event_Count', 'NumMentions': 'Mention_Count'})

# 3. 사건 하나당 평균 기사 수(주목도) 계산
comparison_df['Mentions_per_Event'] = (comparison_df['Mention_Count'] / comparison_df['Event_Count']).round(2)

# 4. 기사량 많은 순으로 정렬하여 출력
print("--- CAMEO 코드별 사건 수 vs 기사 언급량 비교 ---")
display(comparison_df.sort_values(by='Mention_Count', ascending=False))
```

```python
# 1. RootCode 추출 (앞 2자리)
df['RootCode'] = df['EventCode'].astype(str).str.zfill(3).str[:2]

# 2. 분석 대상 고위험군 코드 설정
high_risk_codes = ['15', '19', '20']

# 3. 전체 대비 고위험군 통계 계산
total_event_count = len(df)
total_mention_volume = df['NumMentions'].sum()

high_risk_df = df[df['RootCode'].isin(high_risk_codes)]
high_risk_event_count = len(high_risk_df)
high_risk_mention_volume = high_risk_df['NumMentions'].sum()

# 4. 결과 출력
print(f"--- 중국-대만 고위험군(15, 19, 20) 비중 분석 ---")
print(f"1. 사건 수 기준 (Event Count):")
print(f"   - 고위험군: {high_risk_event_count}건 / 전체: {total_event_count}건")
print(f"   - 비중: {high_risk_event_count / total_event_count * 100:.2f}%")

print(f"\n2. 기사 언급량 기준 (Mention Volume):")
print(f"   - 고위험군: {high_risk_mention_volume:,.0f}건 / 전체: {total_mention_volume:,.0f}건")
print(f"   - 비중: {high_risk_mention_volume / total_mention_volume * 100:.2f}%")
```

```python
import matplotlib.pyplot as plt

# 데이터 준비
labels = ['High-Risk (15,19,20)', 'Others']
event_sizes = [high_risk_event_count, total_event_count - high_risk_event_count]
mention_sizes = [high_risk_mention_volume, total_mention_volume - high_risk_mention_volume]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# 사건 수 비중
ax1.pie(event_sizes, labels=labels, autopct='%1.1f%%', startangle=140, 
        colors=['#e74c3c', '#bdc3c7'], explode=(0.1, 0))
ax1.set_title('Proportion by Event Count (Rows)', fontsize=14, fontweight='bold')

# 기사 언급량 비중
ax2.pie(mention_sizes, labels=labels, autopct='%1.1f%%', startangle=140, 
        colors=['#c0392b', '#ecf0f1'], explode=(0.1, 0))
ax2.set_title('Proportion by Mention Volume (NumMentions)', fontsize=14, fontweight='bold')

plt.suptitle('Market Share of High-Risk Events', fontsize=18, fontweight='bold')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. 기사량 기초 통계량 확인 (상위 % 파악)
print("--- [기사량(NumMentions) 백분위수 분포] ---")
quantiles = df['NumMentions'].quantile([0.5, 0.75, 0.90, 0.95, 0.99])
print(quantiles)

# 2. 기사량 분포 시각화 (너무 극단적인 값이 많아 상위 1%는 제외)
limit_99 = df['NumMentions'].quantile(0.99)
filtered_mentions = df[df['NumMentions'] < limit_99]

plt.figure(figsize=(12, 5))
sns.histplot(filtered_mentions['NumMentions'], bins=50, kde=True, color='skyblue')

# 백분위수 기준선 긋기 (예: 상위 5% 기준선)
threshold_95 = quantiles[0.95]
plt.axvline(threshold_95, color='red', linestyle='--', label=f'Top 5% ({threshold_95:.1f} Mentions)')

plt.title('Distribution of Article Volume (NumMentions)', fontsize=14)
plt.xlabel('Number of Mentions')
plt.ylabel('Frequency (Log Scale)')
plt.yscale('log') # 빈도 차이가 커서 로그 스케일 적용
plt.legend()
plt.show()
```

```python
# 1. Goldstein 기초 통계량 (하위 % 파악 - 점수가 낮을수록 위험)
print("\n--- [GoldsteinScale 백분위수 분포 (위험도 순)] ---")
goldstein_quantiles = df['GoldsteinScale'].quantile([0.01, 0.05, 0.10, 0.25, 0.5])
print(goldstein_quantiles)

# 2. Goldstein 분포 시각화
plt.figure(figsize=(12, 5))
sns.histplot(df['GoldsteinScale'], bins=40, kde=True, color='coral')

# 백분위수 기준선 긋기 (예: 하위 5% 기준선)
threshold_bottom_5 = goldstein_quantiles[0.05]
plt.axvline(threshold_bottom_5, color='darkred', linestyle='--', label=f'Bottom 5% ({threshold_bottom_5:.1f} Score)')
plt.axvline(0, color='black', linestyle=':', alpha=0.5, label='Neutral (0)')

plt.title('Distribution of Conflict Intensity (GoldsteinScale)', fontsize=14)
plt.xlabel('Goldstein Scale (Negative = Conflict)')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
import pandas_gbq
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 날짜 데이터 형식 변환 및 정렬
df['SQLDATE'] = pd.to_datetime(df['SQLDATE'], format='%Y%m%d', errors='coerce')

# 2. 일별 총 기사량(NumMentions) 집계
daily_volume = df.groupby('SQLDATE')['NumMentions'].sum().to_frame()

# 3. 이동평균선(Rolling Mean) 계산
daily_volume['MA_7'] = daily_volume['NumMentions'].rolling(window=7, min_periods=1).mean()
daily_volume['MA_14'] = daily_volume['NumMentions'].rolling(window=14, min_periods=1).mean()
daily_volume['MA_28'] = daily_volume['NumMentions'].rolling(window=28, min_periods=1).mean()

# 4. 시각화(특정 연도만 확대)
start_date = '2025-01-01'
end_date = '2025-12-31'
mask = (daily_volume.index >= start_date) & (daily_volume.index <= end_date)
zoom_df = daily_volume.loc[mask]

plt.figure(figsize=(15, 6))

# 원본 일별 기사량은 옅은 막대그래프로 배경에 둡니다
plt.bar(zoom_df.index, zoom_df['NumMentions'], color='lightgray', alpha=0.5, label='Daily Mentions')

# 이동평균선 그리기
plt.plot(zoom_df.index, zoom_df['MA_7'], color='red', linewidth=2, label='7-Day MA (Fast)')
plt.plot(zoom_df.index, zoom_df['MA_14'], color='orange', linewidth=2, label='14-Day MA (Medium)')
plt.plot(zoom_df.index, zoom_df['MA_28'], color='blue', linewidth=2, label='28-Day MA (Slow)')

plt.title('Daily Article Volume vs. Moving Averages (2022 Zoom-in)', fontsize=14)
plt.xlabel('Date')
plt.ylabel('Total NumMentions')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
