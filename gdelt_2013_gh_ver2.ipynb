```python
import pandas as pd

# 데이터 불러오기 (경로는 본인의 설정에 맞게 수정)
df = pd.read_csv('gdelt_2013_ko.csv')

# 상위 5개 데이터 확인
print(df.head())

# 전체 데이터 정보(행 수, 컬럼 타입 등) 확인
print(df.info())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 상위 10개 사건 코드 시각화
plt.figure(figsize=(10, 6))
df['EventCode'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Event Codes in 2013 (Korea)')
plt.xlabel('Event Code')
plt.ylabel('Frequency')
plt.show()
```

```python
# 날짜를 시계열 데이터로 변환
df['SQLDATE'] = pd.to_datetime(df['SQLDATE'], format='%Y%m%d')

# 날짜별 GoldsteinScale 평균 계산
daily_sentiment = df.groupby('SQLDATE')['GoldsteinScale'].mean()

# 시계열 그래프 그리기
plt.figure(figsize=(12, 5))
daily_sentiment.plot()
plt.axhline(0, color='red', linestyle='--') # 0점 기준선
plt.title('Average Daily Goldstein Scale (2013)')
plt.show()
```
